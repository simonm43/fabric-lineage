```python
# Auto-generated by the Lineage workload item. Do not edit by hand.
CLUSTER_URI = "https://eventhouse-cluster.kusto.fabric.microsoft.com"
DATABASE    = "lineage_eh"
TABLE       = 'lin_deleted_datasets'
MAPPING     = 'lin_deleted_datasets_mapping'
FABRIC_API  = 'https://api.fabric.microsoft.com/v1'
```

```python
import json, datetime, requests
import notebookutils

ws_id = notebookutils.runtime.context.get('currentWorkspaceId')
print(f'Workspace: {ws_id}')

fab_token = notebookutils.credentials.getToken('pbi')
fab_hdr = {'Authorization': f'Bearer {fab_token}'}

lakehouses = requests.get(
    f'{FABRIC_API}/workspaces/{ws_id}/lakehouses',
    headers=fab_hdr, timeout=30,
).json().get('value', [])
print(f"Found {len(lakehouses)} lakehouse(s):", [lh['displayName'] for lh in lakehouses])

live_uris = set()
for lh in lakehouses:
    url = f"{FABRIC_API}/workspaces/{ws_id}/lakehouses/{lh['id']}/tables"
    while url:
        r = requests.get(url, headers=fab_hdr, timeout=30)
        # Some lakehouses (or preview features) may not support the /tables API yet.
        # If the API returns 400/404, log and skip that lakehouse instead of failing the whole job.
        if r.status_code in (400, 404):
            print(f"Skipping lakehouse {lh.get('displayName')} ({lh.get('id')}) - unsupported tables API, status {r.status_code}")
            break
        r.raise_for_status()
        body = r.json()
        for t in body.get('data', []):
            loc = (t.get('location') or '').lower()
            if loc.startswith('abfss://'):
                live_uris.add(loc)
        url = body.get('continuationUri')

print(f'Live tables across all lakehouses: {len(live_uris)}')
```

```python
# Pull every URI that has ever appeared as an OL endpoint, restricted to OneLake.
kusto_token = notebookutils.credentials.getToken('kusto')
kusto_hdr = {'Authorization': f'Bearer {kusto_token}', 'Content-Type': 'application/json'}

kql = (
    'let endpoints = union \n'
    '  (lin_table_edges_mv | project u = tolower(strcat(src_namespace, src_name))),\n'
    '  (lin_table_edges_mv | project u = tolower(strcat(dst_namespace, dst_name))),\n'
    '  (lin_datasets_mv    | project u = tolower(strcat(ds_namespace,  ds_name)));\n'
    "endpoints | where u startswith 'abfss://' and u contains 'onelake' | distinct u"
)
r = requests.post(
    f'{CLUSTER_URI}/v1/rest/query',
    headers=kusto_hdr,
    json={'db': DATABASE, 'csl': kql},
    timeout=60,
)
r.raise_for_status()
ever_seen = {row[0] for row in r.json()['Tables'][0]['Rows']}
print(f'URIs ever seen in OL events: {len(ever_seen)}')
```

```python
tombstones = sorted(ever_seen - live_uris)
print(f'Tombstones (in OL but no longer in any lakehouse): {len(tombstones)}')
for u in tombstones[:10]:
    print('  -', u)
if len(tombstones) > 10:
    print(f'  ... and {len(tombstones) - 10} more')
```

```python
snapshot_id = datetime.datetime.utcnow().isoformat(timespec='seconds') + 'Z'
rows = [{'snapshot_id': snapshot_id, 'full_uri': u, 'snapshot_at': snapshot_id} for u in tombstones]

if not rows:
    # Still emit one marker row so the function returns an empty set, not stale data.
    rows = [{'snapshot_id': snapshot_id, 'full_uri': '', 'snapshot_at': snapshot_id}]

url = (f'{CLUSTER_URI}/v1/rest/ingest/{DATABASE}/{TABLE}'
       f'?streamFormat=MULTIJSON&mappingName={MAPPING}')
body = '\n'.join(json.dumps(r) for r in rows)

resp = requests.post(
    url,
    data=body.encode('utf-8'),
    headers={
        'Authorization': f'Bearer {kusto_token}',
        'Content-Type':  'application/json',
    },
    timeout=30,
)
resp.raise_for_status()
print(f'Ingested tombstone snapshot {snapshot_id} - {len(tombstones)} URI(s) - status {resp.status_code}')
```
